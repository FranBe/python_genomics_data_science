```python
from Bio.Blast import NCBIWWW
from Bio.Blast import NCBIXML
```

```python
# Given DNA sequence
dna_seq = """TGGGCCTCATATTTATCCTATATACCATGTTCGTATGGTGGCGCGATGTTCTACGTGAATCCACGTTCGAAGGACATCATACCAAAGTCGTAC
AATTAGGACCTCGATATGGTTTTATTCTGTTTATCGTATCGGAGGTTATGTTCTTTTTTGCTCTTTTTCGGGCTTCTTCTCATTCTTCTTTGGCAC
CTACGGTAGAG"""
```

```python
# Running BLAST over the internet
# "blastn": program to use
# "nt": wich database to search against
result_handle = NCBIWWW.qblast("blastn", "nt",dna_seq)
```

```python
# As the result is in XML format, another tool could be used to read it
result = NCBIXML.read(result_handle)
```

```python
# The result now, is an object, with different properties. For example:
print(result.descriptions)
print()
print(result.alignments)
```

```python
# Length
len(result.alignments)
```

```python
def GetBlastResults(blast_result:object, E_VALUE_THRESH=0.01):
    """ Get BLAST results from a NCBIXML object

    Args:
        E_VALUE_THRESH(int):
        blast_result(object): NCBIXML read
    Returns:
        
    
    """
    for alignment in blast_result.alignments:
        for hsp in alignment.hsps:
            if hsp.expect < E_VALUE_THRESH:
                print("**** Alignment ****")
                print("sequence: ", alignment.title)
                print("length: ", alignment.length)
                print("e value: ", hsp.expect)
                print(hsp.query)
                print(hsp.match)
                print(hsp.sbjct)
    
```

```python
GetBlastResults(result)
```

### Question 4

According to official documentation and this good guide, ["Reverse Complement of DNA: What It Is and How to Get It"](https://casrai.org/guides/reverse-complement-dna)

|Step|	Result for the input ATGGCGTTAC|
|--|--|
|Input, 5′ to 3′|	ATGGCGTTAC
|Complement only (same order)|	TACCGCAATG
|Reverse only	|CATTGCGGTA
|Reverse complement	|GTAACGCCAT


```python
from Bio.Seq import Seq, reverse_complement

# Method 1: Using Seq object
dna = Seq("ATGCCGTA")
print(dna.reverse_complement())  # Output: TACGGCAT

# Method 2: Using standalone function
print(reverse_complement("ATGCCGTA"))  # Output: TACGGCAT
```
